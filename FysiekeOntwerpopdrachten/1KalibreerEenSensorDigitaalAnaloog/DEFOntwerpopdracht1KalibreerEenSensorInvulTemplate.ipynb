## *Opdracht 6*: Meten en fitten, Leerdoel

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

Temp = np.array([59, 56 ,52 ,49 ,47 ,45 ,43 ,41 ,39 ,37, 35, 33, 31, 29, 27, 25, 23, 21, 19])
bits = np.array([780 ,768 ,720 ,710 ,685 ,670 ,655 ,644 ,624 ,612, 596, 575, 560, 545, 530, 515, 494, 476, 459])
```

```python
def f(x, a, b):
    return a * x + b

# curve fit
popt, pcov = curve_fit(f, Temp, bits)

a_fit, b_fit = popt

print("a =", a_fit)
print("b =", b_fit)

# plot
plt.figure()
plt.scatter(Temp, bits)
plt.plot(Temp, f(Temp, a_fit, b_fit))
plt.xlabel('Temperatuur [$C$]')
plt.ylabel('Bits [$-$]')
plt.show()

bits_30C = f(30, a_fit, b_fit)
bits_40C = f(40, a_fit, b_fit)
Temp_minC = (0-b_fit)/a_fit
Temp_maxC = (1023-b_fit)/a_fit

print(f'aantal bits nodig voor 30 °C: {bits_30C:.0f}')
print(f'aantal bits nodig voor 40 °C: {bits_40C:.0f}')
print(f'de range van de temperatuursensor is: {Temp_minC:.0f} °C tot {Temp_maxC:.0f} °C ')
```
